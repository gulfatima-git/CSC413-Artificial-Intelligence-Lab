## Lab 3 - Task 1
### Represent a small graph problem using Python data structures. Define initial state, goal state and valid actions.


```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': ['G']
}

initial_State = 'A'
goal_State = 'G'

print(f"Initial State: {initial_State}")
print(f"Goal State: {goal_State}")
print(f"Valid Actions: ")
for state in graph:
    print(f"{state} -> {graph[state]}")
```

## Lab 3 - Task 2
### Implement BFS from scratch to find a path from start to goal.


```python
def bfs(graph, start, goal):    #function to find the path from start to goal. It will take graph, starting node, and
    queue = [[start]]           # goal node as input.
    visited = set()
    expanded_Nodes = 0

    while queue:                #Using this queue, we will evaluate whether the current node is our goal node or not.
        path = queue.pop(0)
        current = path[-1]

        if current in visited:  #We are ensuring that if we have already visited the current node then 
            continue            #we should move forward in the graph

        visited.add(current)    #Adding the current node to visited set so we can keep track of it.
        expanded_Nodes += 1

        if current == goal:
            return path, expanded_Nodes

        for neighbor in graph[current]:     #   We are using the for loop to evaluate whether the neighboring node of the 
            if neighbor not in visited:         #current node has already been visited or not. If not, then we are visiting 
                new_Path = path + [neighbor]    #that node next.
                queue.append(new_Path)

    return None, expanded_Nodes


bfs_Path, bfs_Expanded = bfs(graph, initial_State, goal_State)

print("\nBFS Result")
print("Path:", bfs_Path)
print("Path Length:", len(bfs_Path) - 1)
print("Expanded Nodes:", bfs_Expanded)
```

## Lab 3 - Task 3
### Implement DFS for the same problem.

```python
def dfs(graph, start, goal):
    stack = [[start]]
    visited = set()
    expanded_Nodes = 0

    while stack:
        path = stack.pop()
        current = path[-1]

        if current in visited:
            continue

        visited.add(current)
        expanded_Nodes += 1

        if current == goal:
            return path, expanded_Nodes

        for neighbor in reversed(graph[current]):
            if neighbor not in visited:
                new_Path = path + [neighbor]
                stack.append(new_Path)

    return None, expanded_Nodes


dfs_path, dfs_expanded = dfs(graph, initial_State, goal_State)

print("\nDFS Result")
print("Path:", dfs_path)
print("Path Length:", len(dfs_path) - 1)
print("Expanded Nodes:", dfs_expanded)
```
