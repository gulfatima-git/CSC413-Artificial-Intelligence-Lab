## Task 1 - Represent the Problem in Python

```python
graph = {
    'S': {'A': 2, 'B': 1},
    'A': {'G': 10},
    'B': {'C': 2},
    'C': {'G': 2},
    'G': {}
}

heuristic = {
    'S': 0,
    'A': 1,
    'B': 3,
    'C': 2,
    'G': 0
}

start = 'S'
goal = 'G'

print("Neighbors of S:", graph['S'])
print("Heuristic of A:", heuristic['A'])
print("Heuristic of B:", heuristic['B'])
print("Start City:", start)
print("Goal City:", goal)
```

## Task 2 - Implement Greedy Best-First Search

```python
import heapq

def greedy_best_first_search(graph, heuristic, start, goal):
    priority_queue = []
    heapq.heappush(priority_queue, (heuristic[start], start))

    visited = set()
    parent = {start: None}

    while priority_queue:
        h_value, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        for neighbor in graph[current]:
            if neighbor not in visited and neighbor not in parent:
                parent[neighbor] = current
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbor], neighbor)
                )

    return None


def calculate_path_cost(graph, path):
    if path is None:
        return float('inf')

    total_cost = 0

    for i in range(len(path) - 1):
        total_cost += graph[path[i]][path[i + 1]]

    return total_cost


greedy_path = greedy_best_first_search(
    graph, heuristic, start, goal
)

print("Greedy Best-First Search")
print("Path:", greedy_path)
print("Total Cost:", calculate_path_cost(graph, greedy_path))
```

## Task 3 - Implement A* Search

```python
def a_star_search(graph, heuristic, start, goal):
    priority_queue = []
    heapq.heappush(priority_queue, (heuristic[start], start))

    parent = {start: None}
    g_cost = {start: 0}

    while priority_queue:
        f_value, current = heapq.heappop(priority_queue)

        # Ignore outdated queue entries
        if f_value != g_cost[current] + heuristic[current]:
            continue

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()
            return path, g_cost[goal]

        for neighbor, road_cost in graph[current].items():
            new_g = g_cost[current] + road_cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                parent[neighbor] = current

                new_f = new_g + heuristic[neighbor]

                heapq.heappush(
                    priority_queue,
                    (new_f, neighbor)
                )

    return None, float('inf')


astar_path, astar_cost = a_star_search(
    graph, heuristic, start, goal
)

print("A* Search")
print("Path:", astar_path)
print("Total Cost:", astar_cost)
```

## Task 4 - Compare Both Algorithms

```python
greedy_path = greedy_best_first_search(
    graph, heuristic, start, goal
)

greedy_cost = calculate_path_cost(graph, greedy_path)

astar_path, astar_cost = a_star_search(
    graph, heuristic, start, goal
)

print("Algorithm Comparison")
print("-------------------------")
print("GBFS Path:", greedy_path)
print("GBFS Cost:", greedy_cost)

print("A* Path:", astar_path)
print("A* Cost:", astar_cost)

if greedy_cost < astar_cost:
    print("GBFS returned the cheaper route.")
elif astar_cost < greedy_cost:
    print("A* returned the cheaper route.")
else:
    print("Both algorithms returned routes with equal costs.")
```

## Task 6B - Apply A* Search


```python
def a_star_with_expansions(graph, heuristic, start, goal):
    import heapq

    queue = [(heuristic[start], start)]
    g_cost = {start: 0}
    parent = {start: None}
    expansions = []

    while queue:
        f, current = heapq.heappop(queue)

        if f != g_cost[current] + heuristic[current]:
            continue

        expansions.append(
            (current, g_cost[current],
             heuristic[current], f)
        )

        if current == goal:
            path = []
            node = goal

            while node is not None:
                path.append(node)
                node = parent[node]

            return path[::-1], g_cost[goal], expansions

        for neighbor, cost in graph[current].items():
            new_g = g_cost[current] + cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                parent[neighbor] = current
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(queue, (new_f, neighbor))

    return None, float('inf'), expansions


path, cost, expansions = a_star_with_expansions(
    chemical_graph, chemical_heuristic, 'S0', 'G'
)

print("A* Search")
print("Expansion Order:", [x[0] for x in expansions])
print("Optimal Path:", path)
print("Total Cost:", cost)

for node, g, h, f in expansions:
    print(f"{node}: g={g}, h={h}, f={f}")
```

## Task 6C - Apply Iterative Deepening A* (IDA*)

```python
def ida_star(graph, heuristic, start, goal):

    threshold = heuristic[start]
    all_iterations = []

    def dfs(node, g_cost, threshold, path, expanded):
        f_cost = g_cost + heuristic[node]

        if f_cost > threshold:
            return f_cost, None

        expanded.append(node)

        if node == goal:
            return True, path.copy()

        minimum = float('inf')

        for neighbor, cost in graph[node].items():
            if neighbor in path:
                continue

            path.append(neighbor)

            result, solution = dfs(
                neighbor,
                g_cost + cost,
                threshold,
                path,
                expanded
            )

            if result is True:
                return True, solution

            minimum = min(minimum, result)
            path.pop()

        return minimum, None

    while True:
        expanded = []

        result, path = dfs(
            start, 0, threshold, [start], expanded
        )

        all_iterations.append((threshold, expanded.copy()))

        if result is True:
            total_cost = calculate_path_cost(graph, path)
            return path, total_cost, all_iterations

        if result == float('inf'):
            return None, float('inf'), all_iterations

        threshold = result


ida_path, ida_cost, iterations = ida_star(
    chemical_graph, chemical_heuristic, 'S0', 'G'
)

print("IDA* Search")

for i, (threshold, expanded) in enumerate(iterations, 1):
    print(f"Iteration {i}")
    print("Threshold:", threshold)
    print("Expanded:", expanded)
    print()

print("Optimal Path:", ida_path)
print("Total Cost:", ida_cost)
```
